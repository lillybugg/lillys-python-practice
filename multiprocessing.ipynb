```python
#### import libraries
import time
import math
from multiprocessing import Pool

### regular math code
if __name__ == '__main__':
    start = time.perf_counter() #start time
    results1 = [math.factorial(x) for x in range(12000)] #factorial being applied to numbers up to 11,999
    end = time.perf_counter() #end time
    
    print(end-start) #prints how long it takes, it took me almost 11 seconds the first time.

#### We can do this with 5 processes instead!
#### multiprocessing starts here

    start = time.perf_counter() #start time
    with Pool(5) as p: #input processes (don't do more than the number of cores you have in your cpu)
        results2 = p.map(math.factorial, list(range(12000)))
    end = time.perf_counter() #start time

    print(end - start) #took 3.5 seconds first run, 3 seconds with pool 6 (I have 6 cores, but probably shouldnt do idk?)

    print(all(x == y for x, y in zip(results1, results2))) #makes sure results are the same
```

**Pipe example is a communciation channel between processes**

```python
from multiprocessing import Process, Pipe

def worker(conn):
    for i in range(5):
        conn.send(i * i)
    conn.close()

if __name__ == "__main__":
    parent_conn, child_conn = Pipe() #use Pipe(duplex=True) for two-way communication
    p = Process(target=worker, args=(child_conn,))
    p.start()

    for _ in range(5):
        print("Received:", parent_conn.recv())

    p.join()
```
